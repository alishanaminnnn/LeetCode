Given an integer x, return true if x is a palindrome, and false otherwise.

```python
def palindrome(x)->bool:
    reverse=0;
    original=x;
    while original!=0:
        reverse=reverse*10+(original%10)
        original=original//10;
    if reverse==x:
        return True
    else:
        return False

palindrome(12321)

    
```
